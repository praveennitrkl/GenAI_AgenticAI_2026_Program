```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x),color="red")
plt.plot(x, np.cos(x),color="blue")
plt.title("Sine and Cosine Waves")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(["Sine", "Cosine"])

plt.show()
```

```python
# create a plot figure
plt.figure()
x1 = np.linspace(0, 10, 100)

# create the first of two panels and set current axis
plt.subplot(2, 1, 1)   # (rows, columns, panel number)
plt.plot(x1, np.sin(x1),color="red",marker='*');

# create the second of two panels and set current axis
plt.subplot(2, 1, 2)   # (rows, columns, panel number)
plt.plot(x1, np.cos(x1),color="blue",marker='o');

plt.show()
```

```python
x = np.linspace(0, 10, 10)

plt.plot(x, x, color="red")
plt.plot(x, x**2, color="blue")
plt.plot(x, x**3, color="green")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Linear, Quadratic, and Cubic Functions")

plt.legend(["Linear", "Quadratic", "Cubic"])

plt.show()

```

```python
t = np.arange(0., 5., 0.2)
t
```

```python
# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```

```python
# First create a grid of plots
# ax will be an array of two Axes objects
fig, ax = plt.subplots(2)


# Call plot() method on the appropriate object
ax[0].plot(x1, np.sin(x1), 'b-')
ax[1].plot(x1, np.cos(x1), 'b-')
plt.show()
```

```python
fig = plt.figure()

x2 = np.linspace(0, 10, 100)
y2 = x2 ** 2

# Add an axes to the figure
ax = fig.add_axes([1, 1, 1, 0.5])  # [left, bottom, width, height]

ax.plot(x2, y2,c='r')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('Quadratic Function')
plt.show()

```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

ax1.plot(x1, np.sin(x1), 'b-')
ax2.plot(x1, np.cos(x1), 'r-')
ax3.plot(x1, np.tan(x1), 'g-')
ax4.plot(x1, np.exp(x1), 'm-')


fig.show()
```

```python
x4 = range(1, 5)

plt.plot(x4, [xi*1.5 for xi in x4])

plt.plot(x4, [xi*3 for xi in x4])

plt.plot(x4, [xi/3.0 for xi in x4])

plt.show()
```
