```python
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

x1 = np.array([0, 1, 2, 3])
y1 = np.array([2, 9, 1, 15])

x2 = np.array([0, 1, 2, 3])
y2 = np.array([1, 7, 2, 12])

x3 = np.array([0, 1, 2, 3])
y3 = np.array([2, 9, 1, 15])

x4 = np.array([0, 1, 2, 3])
y4 = np.array([1, 6, 3, 14])

x5 = np.array([0, 1, 2, 3])
y5 = np.array([2, 9, 1, 15])


plt.subplot(2, 3, 1)
plt.plot(x,y)

plt.subplot(2, 3, 2)
plt.plot(x1,y1)

plt.subplot(2, 3, 3)
plt.plot(x2,y2)

plt.subplot(2, 3, 4)
plt.plot(x3,y3)

plt.subplot(2, 3, 5)
plt.plot(x4,y4)

plt.subplot(2, 3, 6)
plt.plot(x5,y5)

plt.suptitle("Multiple Plots")

plt.show()
```

Bar Charts

```python
x = np.array(["A", "B", "C", "D","E","F","G","H","I","J","K","L","M"])
y = np.array([3, 8, 1, 10,20,30,25,50,35,70,60,90,75])

plt.bar(x, y,color='m',edgecolor='k',linewidth=1)

plt.xlabel("Categories",c='b')  
plt.ylabel("Values",c='r')
plt.title("Bar Chart",c='g')

plt.show()
```

```python
 x = np.array(["A", "B", "C", "D"])
 y = np.array([3, 8, 1, 10])

 plt.barh(x, y,color='hotpink',edgecolor='k',linewidth=2,height=0.3)
 plt.xlabel("Values",c='b')
 plt.ylabel("Categories",c='r')
 plt.title("Horizontal Bar Chart",c='g')

 plt.show()
```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color = "#4CAF50", width=0.3)
plt.show()
```

Histograms Generation

```python
x = np.random.normal(170, 10, 250)

print(x)

plt.hist(x)
plt.xlabel("Value Ranges", c='b')
plt.ylabel("Frequency", c='r')
plt.title("Histogram", c='g')
plt.show()
```

```python
x = np.random.randint(10, 170, 250)

print(x)

plt.hist(x, bins=20, color='cyan', edgecolor='black', linewidth=1.2)
plt.xlabel("Value Ranges", c='b')
plt.ylabel("Frequency", c='r')
plt.title("Histogram", c='g')
plt.show()
```

Pie Charts


```python
y = np.array([80, 25, 15, 10])
labels = ["Python", "JavaScript", "Java", "C++"]

plt.pie(y, labels=labels,  startangle=180)
plt.title("Programming Language Popularity")
```

```python
y = np.array([80, 25, 15, 10])
labels = ["Python", "JavaScript", "Java", "C++"]

plt.pie(y, labels=labels,  startangle=180,explode=[0.1, 0.1, 0.1, 0.1], shadow=True)
plt.title("Programming Language Popularity")
plt.show()
```
