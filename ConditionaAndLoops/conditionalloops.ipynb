```python
for num in range(1, 11, 3):
    print(num)
```

```python
n = 5
for num in range(2, n):
    if n % num == 0:
        print('Not Prime')
        break
else:
    print('Prime')
```

```python
n = 10
for num in range(2, n):
    if n % num == 0:
        print('Not Prime')
        continue
else:
    print('Prime')
```
